```python
#A)Handling missing values
#1.Mean/Mode/Median Imputation:
import pandas as pd
import numpy as np
#Sample dataset
df=pd.DataFrame({'Age': [25,30,np.nan,40,35],
                 'Department': ['HR','Finance','Finance',np.nan,'IT']})
#Display original dataset
print("Original Dataset(With Missing values):")
print(df)
#Mean for numeric 
df['Age']=df['Age'].fillna(df['Age'].mean())
#Mode for categorical
df['Department']=df['Department'].fillna(df['Department'].mode()[0])
print(df)
```

```python
#2.Forward/Backward filling:
import pandas as pd
import numpy as np
#Sample dataset
df=pd.DataFrame({'Age': [25,30,np.nan,40,35],
                 'Department': ['HR','Finance','Finance',np.nan,'IT']})
#Display original dataset
print("Original Dataset(With Missing values):")
print(df)
#Forward Fill(use previous value)
df_ffill=df.copy()
df_ffill.ffill(inplace=True)
print(df_ffill)
#Backward Fill(use next value)
df_bfill=df.copy()
df_bfill.bfill(inplace=True)
print(df_bfill)
```

```python
#3.a.Delect missing rows/columns:
import pandas as pd
import numpy as np
#Sample dataset
df=pd.DataFrame({'Age': [25,30,np.nan,40,35],
                 'Department': ['HR','Finance','IT','Finance','IT']})
#Display original dataset
print("Original Dataset(With Missing values):")
print(df)
#Drop rows with missing values
df_drop_rows=df.dropna()
print("After dropping rows:\n",df_drop_rows)
#Drop columns with missing values
df_drop_cols=df.dropna(axis=1)
print("After dropping columns:\n",df_drop_cols)
```

```python
#3.b.a Remove duplicates (Exact match removal/direct removal of duplicates)
import pandas as pd
#Sample dataset with duplicates
df=pd.DataFrame({
    'ID':[1,2,2,3,4,4],
    'Name':['Alice','Bob','Bob','Charlie','David','David'],
    'Age':[25,30,30,35,40,40]
})
print("Original Dataset:\n",df)
#remove exact duplicates
df_exact=df.drop_duplicates()
print("\nAfter Exact Match removal:\n",df_exact)    
```

```python
#3.b.b Remove duplicates (based on parameters removal)
import pandas as pd
#Sample dataset with duplicates
df=pd.DataFrame({
    'ID':[1,2,2,3,4,4],
    'Name':['Alice','Bob','Bob','Charlie','David','David'],
    'Age':[25,30,30,35,40,40]
})
print("Original Dataset:\n",df)
#Remove duplicates based only on ID
df_subset_id=df.drop_duplicates(subset=['ID'])
print("\nAfter subset-Based Removal ('ID'):\n",df_subset_id)
#Remove duplicates based only on Name
df_subset_name=df.drop_duplicates(subset=['Name'])
print("\nAfter subset-Based Removal ('Name'):\n",df_subset_name)
```

```python
#3.b.c Remove duplicates (Data standardization) ISO 8601 format(YYYY-MM-DD)
import pandas as pd
#sample dataset
df=pd.DataFrame({
    'Date':['2025-01-05','05/01/2025','Jan 5, 2025','2025.01.05']
})
print("Original Dataset:\n",df)
#convert all dates to standard ISO format(YYYY-MM-DD)
df['Date']=pd.to_datetime(df['Date'],errors='coerce').dt.strftime('%Y-%m-%d')
print("\nAfter formating:")
print(df)
```
